```python
from Bio import SeqIO  
# SeqIo is used to read sequence FASTA file.
Seq = SeqIO.read("PGK1_sequence.fasta", "fasta")
#Read the sequence fasta file and store the complete sequence in Seq variable.
start = 94 
#It is starting position of CCDS where it present.
end = 1347 
# It shows the ending position of CDS.
cds_seq = Seq[start-1:end] 
# Extracts the CDS sequence, start used because python index start from 0.

print("CDS coordinate length:", len(cds_seq)) # Print CDS length coordinating length.
print("Divisible by 3:", len(cds_seq) % 3 == 0 ) # Checking the length is divisible by 3. 
```

```python
database = {
   "TTA": "L", "TTG": "L", "CTT": "L", "CTC": "L", "CTA": "L", "CTG": "L",
    "ATT": "T", "ATC": "I", "ATA": "I",
    "TTT": "F", "TTC": "F",
    "ATG": "M",
    "GTT": "V", "GTC": "V", "GTA": "V", "GTG": "V",
    "TCT": "S", "TCC": "S", "TCA": "S", "TCG": "S", "AGT": "S", "AGC": "S",
    "CCT": "P", "CCC": "P", "CCA": "P", "CCG": "P",
    "ACT": "T", "ACC": "T", "ACA": "T", "ACG": "T",
    "GCT": "A", "GCC": "A", "GCA": "A", "GCG": "A",
    "TAT": "Y", "TAC": "Y",
    "CAT": "H", "CAC": "H",
    "CAA": "Q", "CAG": "Q",
    "AAT": "N", "AAC": "N",
    "AAA": "K", "AAG": "K",
    "GAT": "D", "GAC": "D",
    "GAA": "E", "GAG": "E",
    "TGT": "C", "TGC": "C",
    "TGG": "W",
    "CGT": "R", "CGC": "R", "CGA": "R", "CGG": "R", "AGA": "R", "AGG": "R",
    "GGT": "G", "GGC": "G", "GGA": "G", "GGG": "G",
    "TAA": "*", "TAG": "*", "TGA": "*"
}
# Creating the database containing animo acid codons.
seq_protein = "" 
# Creating an empty string to store translated protein sequence.
for i in range(0, len(cds_seq), 3): # Goes through the CDS sequence 3 nucleotides at a time which called as codon. 
    codon = str(cds_seq.seq[i:i+3]) # Extracting 3 nucleotides from CDS seq and store the codons in codon variable.
    if len(codon) == 3: # It check the extraction has 3 nucleotides.
        amino_acid = database[codon] 
        seq_protein += amino_acid # Adds the amino acid to the protein sequence.

print(seq_protein[:30])
print("Total protein length:", len(seq_protein))
# Prints the first 30 animo acid of translated protein sequence.
```
